```python
# ============================================================
# 00 — Entorno Colab y Drive
# Configuración de carpetas del proyecto
# ============================================================

from pathlib import Path

try:
    from google.colab import drive

    drive.mount('/content/gdrive')

    ROOT = Path(
        '/content/gdrive/MyDrive/Investigacion_Deepfake_Watermarking_2026'
    )

except Exception:
    # Permite ejecutar fuera de Google Colab
    ROOT = Path.cwd() / 'Investigacion_Deepfake_Watermarking_2026'


# ============================================================
# Estructura del proyecto
# ============================================================

DIRS = {

    # --------------------------------------
    # Imágenes reales
    # --------------------------------------
    "real_images":
        ROOT / "datasets/real/lfw/smoke_10/images",

    "real_manifests":
        ROOT / "datasets/real/lfw/smoke_10/manifests",

    # --------------------------------------
    # Imágenes falsas / deepfake
    # Se utilizarán posteriormente
    # --------------------------------------
    "fake_images":
        ROOT / "datasets/fake/smoke_10/images",

    "fake_manifests":
        ROOT / "datasets/fake/smoke_10/manifests",

    # --------------------------------------
    # Datos procesados
    # --------------------------------------
    "processed":
        ROOT / "datasets/processed",

    # --------------------------------------
    # Resultados
    # --------------------------------------
    "results":
        ROOT / "results",

    # --------------------------------------
    # Figuras
    # --------------------------------------
    "figures":
        ROOT / "figures",
}


# Crear las carpetas
for path in DIRS.values():
    path.mkdir(parents=True, exist_ok=True)


print("=" * 70)
print("PROYECTO CONFIGURADO")
print("=" * 70)

print(f"\nROOT:\n{ROOT}\n")

for name, path in DIRS.items():
    print(f"{name:18s}: {path}")
```
